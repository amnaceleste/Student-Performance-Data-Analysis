```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
# Load Dataset
data = {
    "Student": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman",
                "Hina", "Bilal", "Zara", "Hamza", "Noor"],
    "Study_Hours": [2, 5, 3, 6, 4, 7, 2, 8, np.nan, 5],
    "Attendance": [70, 90, 75, 95, 80, 98, 65, 92, 85, np.nan],
    "Marks": [55, 85, 65, 92, 72, 96, 50, 94, 78, 88]
}

df = pd.DataFrame(data)

print("Sample dataset:")
display(df)
```

```python
# Explore dataset
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe())

print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Clean Data

df_clean = df.copy()

# Fill missing numerical values with column averages
df_clean["Study_Hours"] = df_clean["Study_Hours"].fillna(
    df_clean["Study_Hours"].mean()
)
df_clean["Attendance"] = df_clean["Attendance"].fillna(
    df_clean["Attendance"].mean()
)

# Remove duplicate records if any exist
df_clean = df_clean.drop_duplicates()

print("Cleaned dataset:")
display(df_clean)

print("\nMissing values after cleaning:")
print(df_clean.isnull().sum())
```

```python
# Basic Analysis

print("Average marks:", round(df_clean["Marks"].mean(), 2))
print("Highest marks:", df_clean["Marks"].max())
print("Lowest marks:", df_clean["Marks"].min())
print("Average study hours:", round(df_clean["Study_Hours"].mean(), 2))

# Classify students based on marks
df_clean["Performance"] = np.where(
    df_clean["Marks"] >= 80, "High",
    np.where(df_clean["Marks"] >= 60, "Average", "Low")
)

print("\nStudent performance categories:")
display(df_clean[["Student", "Marks", "Performance"]])
```

```python
# Required Charts

plt.figure(figsize=(9, 5))
plt.bar(df_clean["Student"], df_clean["Marks"])
plt.title("Student Marks Comparison")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Line Chart

plt.figure(figsize=(8, 5))
plt.plot(
    df_clean["Student"],
    df_clean["Marks"],
    marker="o"
)
plt.title("Marks Across Students")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Pie Chart

performance_counts = df_clean["Performance"].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(
    performance_counts.values,
    labels=performance_counts.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Student Performance Distribution")
plt.show()
```
